```python
import sympy as sp
from IPython.display import display
def analyze_function(F):
    display(F)
    # Проверка на четность/нечетность 
    if F == F.subs(x, -x):
        print('Функция четная')
    elif F == -F.subs(x, -x):
        print('Функция нечетная')
    else:
        print('Функция общего вида')
    
    # Поиск нулей
    zeroes = sp.solve(F)
    if len(zeroes):
        print('Нули функции:')
        display(zeroes)
    else:
        print('Функция не имеет нулей')
    
    # Промежутки знакопостоянства
    pos = sp.solve_univariate_inequality(F>0, x)
    neg = sp.solve_univariate_inequality(F<0, x)
    if pos != False:
        print('Функция положительна:')
        display(pos)
    if neg != False:
        print('Функция отрицательна:')
        display(neg)
    
    # График
    sp.plot(F)
```

```python
# Исправленная функция
import sympy as sp
from IPython.display import display

def analyze_function(F):
    display(F)
    
    # Упрощаем выражения для корректного сравнения
    F_x = sp.simplify(F)
    F_minus_x = sp.simplify(F.subs(x, -x))
    
    # Проверка на четность/нечетность
    if sp.simplify(F_x - F_minus_x) == 0:
        print('Функция четная')
    elif sp.simplify(F_x + F_minus_x) == 0:
        print('Функция нечетная')
    else:
        print('Функция общего вида')
    
    # Поиск нулей
    try:
        zeroes = sp.solve(F, x)
        if zeroes:
            print('Нули функции:')
            display(zeroes)
        else:
            print('Функция не имеет нулей')
    except:
        print('Не удалось найти нули аналитически')
    
    # Промежутки знакопостоянства
    try:
        pos = sp.solve_univariate_inequality(F > 0, x)
        if pos:
            print('Функция положительна:')
            display(pos)
    except:
        pass
    
    try:
        neg = sp.solve_univariate_inequality(F < 0, x)
        if neg:
            print('Функция отрицательна:')
            display(neg)
    except:
        pass
    
    # График
    sp.plot(F, title=str(F), show=True)
```

```python
# Проверка
x = sp.symbols('x')

print("=" * 50)
F1 = 5*x + 1.5 
analyze_function(F1)

print("\n" + "=" * 50)
F2 = 2 * sp.exp(0.2*x) - 1
analyze_function(F2)

print("\n" + "=" * 50)
F3 = 3 * sp.sqrt(x) - 15
analyze_function(F3)

print("\n" + "=" * 50)
F4 = 2 * sp.ln(x) + 5
analyze_function(F4)

print("\n" + "=" * 50)
F5 = -5 * x**2 - 10 * x + 100
analyze_function(F5)
```

```python
import sympy as sp

x = sp.symbols('x')

def check_parity(F):
    F_x = sp.simplify(F)
    F_minus_x = sp.simplify(F.subs(x, -x))
    
    is_even = sp.simplify(F_x - F_minus_x) == 0
    is_odd = sp.simplify(F_x + F_minus_x) == 0
    
    if is_even:
        return "чётная"
    elif is_odd:
        return "нечётная"
    else:
        return "общего вида (ни чётная, ни нечётная)"

# Проверяем функции
funcs = {
    "F2 = 2·exp(0.2x) - 1": 2 * sp.exp(0.2*x) - 1,
    "F3 = 3·√x - 15": 3 * sp.sqrt(x) - 15,
    "F4 = 2·ln(x) + 5": 2 * sp.ln(x) + 5,
    "F5 = -5x² - 10x + 100": -5 * x**2 - 10 * x + 100
}

for name, f in funcs.items():
    print(f"{name}: {check_parity(f)}")
```

$F(−x)=−F(x)$

Вычисляем $F(−x)$:
$$F(−x)=3⋅(−x)−9=−3x−9$$

Сравниваем с $F(x)=3x−9$:
$$−3x−9 = 3x−9$$
$ при: x=1$
$$-12\neq-6$$

Сравниваем с $-F(x)=-3x+9$:
$$−3x−9 = -3x+9$$
Вычитаем $−3x:$
$$-9\neq9$$

2. Нули функции
Решаем уравнение $F(x)=0$:

$$3x−9=0$$
$$3x=9$$
$$x=3$$
Ответ: 
$x=3$

3. Промежутки знакопостоянства
Решаем неравенства.

$F(x)>0:$
$$3x−9>0$$
$$3x>9$$
$$x>3$$

Ответ: $(3,+\infty)$

$F(x)<0:$
$$3x−9<0$$
$$3x<9$$
$$x<3$$

Ответ: $(-\infty, 3)$

4. Промежутки возрастания / убывания
Находим производную:
$$F′(x)=3$$

Знак производной:
$$F′(x)=3>0 для всех x$$

Правило:
- F′(x)>0 → функция возрастает
- F′(x)<0 → функция убывает

Ответ: функция возрастает на всей области определения 
$(-\infty, +\infty)$

```python
from sympy import symbols, simplify, solve, solve_univariate_inequality, diff, plot

x = symbols('x')
F = 3*x - 9


print("Чётность:")
F_minus_x = F.subs(x, -x)
if simplify(F - F_minus_x) == 0:
    print("  Чётная")
elif simplify(F + F_minus_x) == 0:
    print("  Нечётная")
else:
    print("  Ни чётная, ни нечётная")

print("\nНули:", solve(F, x))

print("\nF(x) > 0:", solve_univariate_inequality(F > 0, x))
print("F(x) < 0:", solve_univariate_inequality(F < 0, x))

print("\nПроизводная:", diff(F, x))
print("Возрастает:", solve_univariate_inequality(diff(F, x) > 0, x))
print("Убывает:", solve_univariate_inequality(diff(F, x) < 0, x))

plot(F, (x, -5, 10), 
     title='График функции F(x) = 3x - 9',
     xlabel='x', 
     ylabel='F(x)',
     line_color='blue');
```

#### Задание 5
Напишите функцию analyze_function(F, x), которая:

1. Выводит исходную функцию
2. Определяет чётность / нечётность
3. Находит нули функции
4. Определяет промежутки знакопостоянства
5. Находит промежутки возрастания и убывания (через производную)
6. Строит график функции

Функция должна работать для любого символьного выражения.

```python
import sympy as sp
from IPython.display import display

def analyze_function(F, x):
    """
    Полное исследование функции F(x)
    """
    print("=" * 50)
    print("Исследуемая функция:")
    display(F)

    # 1. Чётность/нечётность
    F_minus_x = F.subs(x, -x)
    if sp.simplify(F - F_minus_x) == 0:
        print("Функция чётная")
    elif sp.simplify(F + F_minus_x) == 0:
        print("Функция нечётная")
    else:
        print("Функция ни чётная, ни нечётная")

    # 2. Нули функции
    zeros = sp.solve(F, x)
    if zeros:
        print("Нули функции:", zeros)
    else:
        print("Функция не имеет нулей")

    # 3. Знакопостоянство
    try:
        positive = sp.solve_univariate_inequality(F > 0, x)
        print("F(x) > 0:", positive)
    except:
        print("F(x) > 0: не удалось определить аналитически")

    try:
        negative = sp.solve_univariate_inequality(F < 0, x)
        print("F(x) < 0:", negative)
    except:
        print("F(x) < 0: не удалось определить аналитически")

    # 4. Монотонность
    derivative = sp.diff(F, x)
    print("Производная:", derivative)

    try:
        increasing = sp.solve_univariate_inequality(derivative > 0, x)
        print("Возрастает:", increasing)
    except:
        print("Возрастает: не удалось определить аналитически")

    try:
        decreasing = sp.solve_univariate_inequality(derivative < 0, x)
        print("Убывает:", decreasing)
    except:
        print("Убывает: не удалось определить аналитически")

    # 5. График
    sp.plot(F, title=str(F), show=True)

```
