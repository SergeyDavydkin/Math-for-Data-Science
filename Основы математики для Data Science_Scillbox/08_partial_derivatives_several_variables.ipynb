### Урок 5. Практика. Частные производные и минимизация MSE (часть 2).


Выпишите MSE, найдите его частные производные, затем с помощью sympy найдите точку минимума и предскажите площадь дома ценой 666 тысяч рублей и высотой в 4 этажа. Получилось ли у вас значение цены, близкое к тому, которое было нам известно на уроке?

|цена дома | количество этажей | площадь дома
|----------|-------------------|-------------
| 200      | 2                 | 200
| 300      | 1                 | 400
| 600      | 3                 | 550
| 666      | 4                 | ?

#### Решение задачи

**Шаг 1. Запись модели**

Будем искать линейную зависимость площади дома $y$ от цены $x_1$ и количества этажей $x_2$
$$ y = a_0 + a_1x_1 + a_2x_2 $$
где:
* $y$ - площадь дома
* $x_1$ - цена дома
* $x_2$ - количество этажей

**Шаг 2. Данные**

| № |цена дома $x_1$ | количество этажей $x_2$ | площадь дома $y$
|---|----------------|-------------------------|-------------
| 1 | 200            | 2                       | 200
| 2 | 300            | 1                       | 400
| 3 | 600            | 3                       | 550

**Шаг 3. MSE (среднеквадратичная ошибка)**

$$ MSE(a_0, a_1, a_2) = \frac{1}{3}\displaystyle\sum_{k=1}^3 (a_0 + a_1x_{1i} + a_2x_{2i} - y_i)^2 $$

Подставляем данные:
$$ MSE = \frac{1}{3}((a_0 + 200a_1 + 2a_2 - 200)^2 + (a_0 + 300a_1 + 1a_2 - 400)^2 + (a_0 + 600a_1 + 3a_2 - 550)^2) $$



**Шаг 4. Нахождение частных производных в SymPy**

```python
import sympy as sp

# Переменные
a0, a1, a2 = sp.symbols('a0 a1 a2')

# MSE (без множителя 1/3 для упрощения)
MSE = (a0 + 200*a1 + 2*a2 - 200)**2 \
    + (a0 + 300*a1 + 1*a2 - 400)**2 \
    + (a0 + 600*a1 + 3*a2 - 550)**2

MSE = MSE / 3

# Частные производные
dMSE_da0 = sp.diff(MSE, a0)
dMSE_da1 = sp.diff(MSE, a1)
dMSE_da2 = sp.diff(MSE, a2)

print("∂MSE/∂a0 =", sp.simplify(dMSE_da0))
print("∂MSE/∂a1 =", sp.simplify(dMSE_da1))
print("∂MSE/∂a2 =", sp.simplify(dMSE_da2))
```

**Шаг 5. Решение системы уравнений (точка минимума)**

```python
# Система уравнений
eq1 = sp.Eq(dMSE_da0, 0)
eq2 = sp.Eq(dMSE_da1, 0)
eq3 = sp.Eq(dMSE_da2, 0)

# Решение
solution = sp.solve([eq1, eq2, eq3], (a0, a1, a2))
print("Решение:")
print(f"a0 = {solution[a0]}")
print(f"a1 = {solution[a1]}")
print(f"a2 = {solution[a2]}")
```

**Шаг 7. Предсказание для дома с ценой 666 и высотой 4 этажа**

$$ y = 160 + 11/10 \sdot 666 + (-90) \sdot 4 = 532.6 $$

```python
import sympy as sp

# Переменные
a0, a1, a2 = sp.symbols('a0 a1 a2')

# Данные
data = [
    (200, 2, 200),
    (300, 1, 400),
    (600, 3, 550)
]

# MSE
MSE = 0
for x1, x2, y in data:
    MSE += (a0 + a1*x1 + a2*x2 - y)**2
MSE = MSE / len(data)

# Частные производные
dMSE_da0 = sp.diff(MSE, a0)
dMSE_da1 = sp.diff(MSE, a1)
dMSE_da2 = sp.diff(MSE, a2)

# Решение системы
solution = sp.solve([dMSE_da0, dMSE_da1, dMSE_da2], (a0, a1, a2))

a0_opt = solution[a0]
a1_opt = solution[a1]
a2_opt = solution[a2]

print(f"a0 = {a0_opt}")
print(f"a1 = {a1_opt}")
print(f"a2 = {a2_opt}")

# Предсказание
x1_new, x2_new = 666, 4
y_pred = a0_opt + a1_opt*x1_new + a2_opt*x2_new
print(f"Предсказанная площадь: {sp.simplify(y_pred)}")
```

### Задание 1. Частные производные функции и их графики

Найдите частные производные функций. Постройте графики этих производных и найдите их нули:

Задание 1 (а)

$$
f(x,y) = x(y - 5)^2
$$

1. Частные производные

По переменной x:
$$
\frac{\partial f}{\partial x} = (y-5)^2 \sdot \frac{\partial x}{\partial x} = (y-5)^2
$$

По переменной y:
$$
\frac{\partial f}{\partial y} = x \sdot \frac{\partial }{\partial y} (y-5)^2 =x \sdot 2(y-5) \sdot 1 = 2x(y-5)
$$

```python
from sympy import *
from sympy.plotting import plot3d
from IPython.display import display, Math
init_printing(use_unicode=False, wrap_line=False, no_global=True)

# 1. Определяем переменные и функцию
x, y = symbols('x, y')
f = x * (y - 5)**2

print("=" * 50)
print("ИСХОДНАЯ ФУНКЦИЯ")
print("=" * 50)
display(Math(f"f(x,y) = {latex(f)}"))

# 2. Частные производные
fx = diff(f, x)
fy = diff(f, y)

print("=" * 50)
print("ЧАСТНЫЕ ПРОИЗВОДНЫЕ")
print("=" * 50)
display(Math(f"∂f/∂x = {latex(fx)}"))
display(Math(f"∂f/∂y = {latex(fy)}"))

# 3. Нули частных производных
print("\n" + "=" * 50)
print("НУЛИ ПРОИЗВОДНЫХ")
print("=" * 50)

# Нули ∂f/∂x
zeros_x = solve(fx, y)
# print(f"∂f/∂x = 0 → y = {zeros_x}")
display(Math(f"∂f/∂x = 0 → y = {latex(zeros_x)}"))

# Нули ∂f/∂y
zeros_y = solve(fy, (x, y))
# print(f"∂f/∂y = 0 → {zeros_y}")
display(Math(f"∂f/∂y = 0 → {latex(zeros_y)}"))

# 4. Графики производных
print("\n" + "=" * 50)
print("ГРАФИКИ")
print("=" * 50)

# График ∂f/∂x
plot(fx, (y, -5, 15), title='∂f/∂x = (y-5)²', xlabel='y', ylabel='∂f/∂x');

# 3D график ∂f/∂y
plot3d(fy, (x, -10, 10), (y, -5, 15), title='∂f/∂y = 2x(y-5)');
```

### Задание 2. Минимизация среднеквадратической ошибки

Найдите наилучшую с точки зрения среднеквадратической ошибки функцию, приближающую зависимость времени разгона автомобиля до 100 км/ч от его массы и мощности двигателя. 

Данные для аппроксимации — в таблице.

| Масса | Мощность двгателя | Время разгона
|-------|-------------------|--------------
| 1945  | 560               | 4.3
| 1495  | 340               | 4.9
| 1570  | 343               | 5.2
| 1520  | 431               | ?

Выпишите MSE, найдите её частные производные, затем с помощью SymPy найдите точку минимума и предскажите время разгона до 100 км/ч автомобиля с максимальной мощностью двигателя 431 л. с. и массой 1520 кг.

**Шаг 1. Модель**

Ищем линейную зависимость времени разгона $y$ от массы $x_1$ и мощности двигателя $x_2$:
$$
y = a_0 + a_1x_1 + a_2x_2
$$

где:

* $y$ - время разгона до 100 км/ч (в секундах)
* $x_1$ - масса автомобиля (в кг)
* $x_2$ - мощность двигателя (в л.с.)

**Шаг 2. Данные**

| № | Масса | Мощность двгателя | Время разгона
|---|-------|-------------------|--------------
| 1 | 1945  | 560               | 4.3
| 2 | 1495  | 340               | 4.9
| 3 | 1570  | 343               | 5.2

**Шаг 3. MSE (среднеквадратичная ошибка)**

$$ MSE(a_0, a_1, a_2) = \frac{1}{3}\displaystyle\sum_{k=1}^3 (a_0 + a_1x_{1i} + a_2x_{2i} - y_i)^2 $$

Подставляем данные:
$$ MSE = \frac{1}{3}((a_0 + 1945a_1 + 560a_2 - 4.3)^2 + (a_0 + 1495a_1 + 340a_2 - 4.9)^2 + (a_0 + 1570a_1 + 343a_2 - 5.2)^2) $$

**Шаг 4. Решение в SymPy**

```python
import sympy as sp
from sympy import symbols, diff, solve, expand, simplify, Matrix, lambdify
import numpy as np

# Переменные
a0, a1, a2 = symbols('a0 a1 a2')

# Данные: (масса, мощность, время)
data = [
    (1945, 560, 4.3),
    (1495, 340, 4.9),
    (1570, 343, 5.2)
]

# Функция потерь (MSE)
MSE = 0
for x1, x2, y in data:
    MSE += (a0 + a1*x1 + a2*x2 - y)**2
MSE = MSE / len(data)

print("=" * 60)
print("MSE")
print("=" * 60)
print(f"MSE = {MSE}")

# ============================================
# Частные производные
# ============================================
print("\n" + "=" * 60)
print("ЧАСТНЫЕ ПРОИЗВОДНЫЕ")
print("=" * 60)

dMSE_da0 = diff(MSE, a0)
dMSE_da1 = diff(MSE, a1)
dMSE_da2 = diff(MSE, a2)

print(f"∂MSE/∂a0 = {simplify(dMSE_da0)}")
print(f"∂MSE/∂a1 = {simplify(dMSE_da1)}")
print(f"∂MSE/∂a2 = {simplify(dMSE_da2)}")

# ============================================
# Решение системы (точка минимума)
# ============================================
print("\n" + "=" * 60)
print("ТОЧКА МИНИМУМА")
print("=" * 60)

solution = solve([dMSE_da0, dMSE_da1, dMSE_da2], (a0, a1, a2))
a0_opt = solution[a0]
a1_opt = solution[a1]
a2_opt = solution[a2]

print(f"a0 = {a0_opt.evalf():.6f}")
print(f"a1 = {a1_opt.evalf():.6f}")
print(f"a2 = {a2_opt.evalf():.6f}")

# ============================================
# Модель
# ============================================
print("\n" + "=" * 60)
print("МОДЕЛЬ")
print("=" * 60)

model = a0_opt + a1_opt*x1 + a2_opt*x2
print(f"y = {a0_opt.evalf():.6f} + {a1_opt.evalf():.6f}·x1 + {a2_opt.evalf():.6f}·x2")

# ============================================
# Предсказание
# ============================================
print("\n" + "=" * 60)
print("ПРЕДСКАЗАНИЕ")
print("=" * 60)

# Автомобиль с массой 1520 кг и мощностью 431 л.с.
x1_new, x2_new = 1520, 431

y_pred = a0_opt + a1_opt*x1_new + a2_opt*x2_new
y_pred_num = y_pred.evalf()

print(f"Для автомобиля с массой {x1_new} кг и мощностью {x2_new} л.с.:")
print(f"Предсказанное время разгона: {y_pred_num:.4f} сек")
```

**Шаг 5. Проверка на исходных данных**

```python
print("\nПроверка на исходных данных:")
for x1, x2, y_true in data:
    y_pred = a0_opt + a1_opt*x1 + a2_opt*x2
    print(f"Масса={x1}, Мощность={x2} → предсказано {y_pred.evalf():.2f} сек, фактически {y_true} сек")
```

```python
from sympy import *
from IPython.display import display, Math
import numpy as np

init_printing(use_unicode=False, wrap_line=False, no_global=True)

def linear_regression(data, x1_new=None, x2_new=None, verbose=True):
    """
    Универсальная линейная регрессия y = a0 + a1*x1 + a2*x2
    """
    # Символьные переменные
    a0, a1, a2 = symbols('a0 a1 a2', real=True)
    x1, x2 = symbols('x1 x2', real=True)
    
    # 1. Строим MSE (автоматически для любого количества точек)
    n = len(data)
    MSE = sum((a0 + a1*x1i + a2*x2i - yi)**2 for x1i, x2i, yi in data) / n
    
    # 2. Частные производные
    grad = [diff(MSE, var) for var in (a0, a1, a2)]
    
    # 3. Решение системы (универсально!)
    try:
        sol = solve(grad, (a0, a1, a2), dict=True)
        if sol and len(sol) > 0:
            sol_dict = sol[0]
            a0_opt = sol_dict[a0]
            a1_opt = sol_dict[a1]
            a2_opt = sol_dict[a2]
            method = "ТОЧНОЕ РЕШЕНИЕ (sympy)"
        else:
            raise ValueError("Нет решения")
    except:
        # Если не получилось — используем численное решение
        X = np.array([[1, x1i, x2i] for x1i, x2i, _ in data], dtype=float)
        y = np.array([yi for _, _, yi in data], dtype=float)
        coeffs, _, _, _ = np.linalg.lstsq(X, y, rcond=None)
        a0_opt, a1_opt, a2_opt = coeffs
        sol_dict = {a0: a0_opt, a1: a1_opt, a2: a2_opt}
        method = "ЧИСЛЕННОЕ РЕШЕНИЕ (numpy)"
    
    # 4. Модель
    model = a0_opt + a1_opt*x1 + a2_opt*x2
    
    # 5. Вывод
    if verbose:
        print("=" * 60)
        print(f"ЛИНЕЙНАЯ РЕГРЕССИЯ ({method})")
        print("=" * 60)
        print(f"Количество точек: {n}")
        
        # Вывод MSE
        display(Math(f"MSE = {latex(MSE)}"))
        
        # Частные производные
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_0}} = {latex(simplify(grad[0]))}"))
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_1}} = {latex(simplify(grad[1]))}"))
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_2}} = {latex(simplify(grad[2]))}"))
        
        # Коэффициенты
        display(Math(f"y = {latex(model.evalf() if method != 'ТОЧНОЕ РЕШЕНИЕ (sympy)' else model)}"))
        display(Math(f"a_0 = {float(a0_opt):.6f}, a_1 = {float(a1_opt):.6f}, a_2 = {float(a2_opt):.6f}"))
        
        # MSE в точке минимума
        y_pred = np.array([float(a0_opt + a1_opt*x1i + a2_opt*x2i) for x1i, x2i, _ in data])
        y_true = np.array([yi for _, _, yi in data])
        mse = np.mean((y_true - y_pred)**2)
        display(Math(f"MSE_{{\\text{min}}} = {mse:.6f}"))
        
        # Проверка
        print("\nПроверка на исходных данных:")
        for x1i, x2i, yi in data:
            y_pred_i = float(a0_opt + a1_opt*x1i + a2_opt*x2i)
            print(f"  x1={x1i:4d}, x2={x2i:3d} → y={y_pred_i:.4f}, факт={yi:.1f}, ошибка={abs(y_pred_i-yi):.4f}")
        
        # Предсказание
        if x1_new is not None and x2_new is not None:
            y_pred_new = float(a0_opt + a1_opt*x1_new + a2_opt*x2_new)
            print(f"\nПредсказание для x1={x1_new}, x2={x2_new}:")
            display(Math(f"\\hat{{y}} = {y_pred_new:.4f}"))
        
        print("=" * 60)
    
    return model, sol_dict
```

```python
# применение универсального решения в Задании 2
# Данные: (масса, мощность, время)
data = [
    (1945, 560, 4.3),
    (1495, 340, 4.9),
    (1570, 343, 5.2)
]

# Автомобиль с массой 1520 кг и мощностью 431 л.с.
x1_new, x2_new = 1520, 431
model, sol = linear_regression(data, x1_new=x1_new, x2_new=x2_new, verbose=True)
```

```python
from sympy import *
from sympy.plotting import plot3d, plot_implicit
from IPython.display import display, Math
import numpy as np
import matplotlib.pyplot as plt

init_printing(use_unicode=False, wrap_line=False, no_global=True)

# ============================================
# АВТОМАТИЧЕСКИЙ TITLE ДЛЯ ГРАФИКОВ
# ============================================
def get_title(expr, var=None):
    """Создаёт LaTeX-заголовок для выражения"""
    if var == 'fx':
        return f'$\\partial f/\\partial x = {latex(expr)}$'
    elif var == 'fy':
        return f'$\\partial f/\\partial y = {latex(expr)}$'
    elif var == 'zeros':
        return f'Нули: ${latex(expr)} = 0$'
    else:
        return f'${latex(expr)}$'

# ============================================
# УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ДЛЯ КОНТУРНЫХ ГРАФИКОВ
# ============================================
def plot_contour_derivative(expr, x, y, x_range=(-5, 5), y_range=(-5, 5), 
                            title=None, ax=None, colorbar=True, levels=20):
    """
    Строит контурный график для символьного выражения (производной).
    """
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 6))
    
    # Преобразуем в числовую функцию
    expr_num = lambdify((x, y), expr, 'numpy')
    
    # Сетка
    X, Y = np.meshgrid(
        np.linspace(x_range[0], x_range[1], 100),
        np.linspace(y_range[0], y_range[1], 100)
    )
    
    # Вычисляем значения
    try:
        Z = expr_num(X, Y)
    except:
        # Если возникла ошибка (например, деление на ноль)
        Z = np.zeros_like(X)
        print(f"Предупреждение: ошибка при вычислении {expr}")
    
    # Контурный график
    contour = ax.contourf(X, Y, Z, levels=levels, cmap='coolwarm')
    ax.contour(X, Y, Z, levels=[0], colors='red', linewidths=2)
    
    # Заголовок
    if title is None:
        title = f'${latex(expr)}$'
    ax.set_title(title, fontsize=12)
    ax.set_xlabel('x', fontsize=10)
    ax.set_ylabel('y', fontsize=10)
    
    if colorbar:
        plt.colorbar(contour, ax=ax)
    
    return ax


# ============================================
# 1. Определение функции
# ============================================
x, y = symbols('x y')
f = x**2 * y + 3 * y**2

display(Math(f"f(x,y) = {latex(f)}"))

# ============================================
# 2. Частные производные
# ============================================
fx = diff(f, x)
fy = diff(f, y)

display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = {latex(fx)}"))
display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = {latex(fy)}"))

# ============================================
# 3. Нули частных производных
# ============================================
print("\n" + "=" * 60)
print("НУЛИ ЧАСТНЫХ ПРОИЗВОДНЫХ")
print("=" * 60)

zeros_x = solve(fx, (x, y))
zeros_y = solve(fy, (x, y))

display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = 0 \\rightarrow {latex(zeros_x)}"))
display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = 0 \\rightarrow {latex(zeros_y)}"))

# ============================================
# 4. 3D графики частных производных (с автоматическим title)
# ============================================
print("\n" + "=" * 60)
print("3D ГРАФИКИ ЧАСТНЫХ ПРОИЗВОДНЫХ")
print("=" * 60)

plot3d(fx, (x, -5, 5), (y, -5, 5),
       title=get_title(fx, 'fx'),
       xlabel='x', ylabel='y', zlabel='∂f/∂x')

plot3d(fy, (x, -5, 5), (y, -5, 5),
       title=get_title(fy, 'fy'),
       xlabel='x', ylabel='y', zlabel='∂f/∂y')

# ============================================
# 5. Графики нулей (с автоматическим title)
# ============================================
print("\n" + "=" * 60)
print("ГРАФИКИ НУЛЕЙ НА ПЛОСКОСТИ")
print("=" * 60)

plot_implicit(Eq(fx, 0), (x, -5, 5), (y, -5, 5),
              title=get_title(fx, 'zeros'),
              xlabel='x', ylabel='y')

plot_implicit(Eq(fy, 0), (x, -5, 5), (y, -5, 5),
              title=get_title(fy, 'zeros'),
              xlabel='x', ylabel='y')


# ============================================
# 6. Контурные графики (matplotlib)
# ============================================
print("\n" + "=" * 60)
print("КОНТУРНЫЕ ГРАФИКИ")
print("=" * 60)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Автоматическое построение без ручного Z_dx, Z_dy
plot_contour_derivative(fx, x, y, ax=axes[0])
plot_contour_derivative(fy, x, y, ax=axes[1])

plt.tight_layout()
plt.show()
```

```python
from sympy import *
from sympy.plotting import plot3d, plot_implicit
from IPython.display import display, Math
import numpy as np
import matplotlib.pyplot as plt

init_printing(use_unicode=False, wrap_line=False, no_global=True)

# ============================================
# АВТОМАТИЧЕСКОЕ ОПРЕДЕЛЕНИЕ ОДЗ
# ============================================
def get_domain(expr, variables):
    """
    Определяет область определения символьного выражения.
    Возвращает условия в виде строки.
    """
    from sympy import log, sqrt, asin, acos, atan, ln, pi
    
    domain_conditions = []
    
    # 1. Проверяем логарифмы: аргумент > 0
    for atom in expr.atoms(log):
        arg = atom.args[0]
        domain_conditions.append(f"{latex(arg)} > 0")
    
    # 2. Проверяем корни чётной степени: подкоренное выражение >= 0
    for atom in expr.atoms(Pow):
        if atom.exp == 1/2:  # sqrt
            arg = atom.base
            domain_conditions.append(f"{latex(arg)} \\ge 0")
    
    # 3. Проверяем знаменатели: != 0
    for atom in expr.atoms(Pow):
        if atom.exp < 0:  # 1/x
            arg = atom.base
            domain_conditions.append(f"{latex(arg)} \\ne 0")
    
    # 4. Проверяем дроби
    if expr.is_Mul or expr.is_Add:
        for arg in expr.args:
            if arg.is_Pow and arg.exp < 0:
                domain_conditions.append(f"{latex(arg.base)} \\ne 0")
    
    # 5. Для обратных тригонометрических функций
    for atom in expr.atoms(asin, acos):
        arg = atom.args[0]
        domain_conditions.append(f"-1 \\le {latex(arg)} \\le 1")
    
    # 6. Для тангенса: cos(x) != 0
    for atom in expr.atoms(tan):
        arg = atom.args[0]
        domain_conditions.append(f"\\cos({latex(arg)}) \\ne 0")
    
    # 7. Для арктангенса: знаменатель != 0
    for atom in expr.atoms(atan):
        arg = atom.args[0]
        if arg.is_Mul or arg.is_Pow:
            # Проверяем знаменатель в аргументе
            for sub_atom in arg.atoms(Pow):
                if sub_atom.exp < 0:
                    domain_conditions.append(f"{latex(sub_atom.base)} \\ne 0")
    
    # Убираем дубликаты
    domain_conditions = list(set(domain_conditions))
    
    if not domain_conditions:
        return "\\mathbb{R}^2"
    
    return " \\text{ и } ".join(domain_conditions)


# ============================================
# ФУНКЦИЯ ДЛЯ ОПРЕДЕЛЕНИЯ ОГРАНИЧЕНИЙ ГРАФИКОВ
# ============================================
def get_graph_limits(domain_conditions, variables, default_range=(-5, 5)):
    """
    Определяет ограничения для графиков на основе ОДЗ.
    """
    x, y = variables
    limits = {}
    
    for cond in domain_conditions:
        if "y" in cond:
            if "> 0" in cond:
                limits['y_min'] = 0.1
            elif "\\ne 0" in cond:
                limits['y_min'] = -0.1
                limits['y_max'] = 0.1
        if "x" in cond:
            if "> 0" in cond:
                limits['x_min'] = 0.1
            elif "\\ne 0" in cond:
                limits['x_min'] = -0.1
                limits['x_max'] = 0.1
    
    return limits


# ============================================
# АВТОМАТИЧЕСКИЙ TITLE ДЛЯ ГРАФИКОВ
# ============================================
def get_title(expr, var=None):
    if var == 'fx':
        return f'$\\partial f/\\partial x = {latex(expr)}$'
    elif var == 'fy':
        return f'$\\partial f/\\partial y = {latex(expr)}$'
    elif var == 'zeros':
        return f'Нули: ${latex(expr)} = 0$'
    else:
        return f'${latex(expr)}$'


# ============================================
# УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ДЛЯ КОНТУРНЫХ ГРАФИКОВ
# ============================================
def plot_contour_derivative(expr, x, y, x_range=(-5, 5), y_range=(-5, 5),
                            title=None, ax=None, colorbar=True, levels=20):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 6))
    
    # Преобразуем в числовую функцию
    expr_num = lambdify((x, y), expr, 'numpy')
    
    # Сетка
    X, Y = np.meshgrid(
        np.linspace(x_range[0], x_range[1], 100),
        np.linspace(y_range[0], y_range[1], 100)
    )
    
    try:
        Z = expr_num(X, Y)
    except:
        Z = np.zeros_like(X)
        print(f"Предупреждение: ошибка при вычислении {expr}")
    
    # Контурный график
    contour = ax.contourf(X, Y, Z, levels=levels, cmap='coolwarm')
    ax.contour(X, Y, Z, levels=[0], colors='red', linewidths=2)
    
    if title is None:
        title = f'${latex(expr)}$'
    ax.set_title(title, fontsize=12)
    ax.set_xlabel('x', fontsize=10)
    ax.set_ylabel('y', fontsize=10)
    
    if colorbar:
        plt.colorbar(contour, ax=ax)
    
    return ax


# ============================================
# ОСНОВНАЯ ФУНКЦИЯ ДЛЯ АНАЛИЗА
# ============================================
def analyze_function(f, x, y, x_range=(-5, 5), y_range=(-5, 5)):
    """
    Полный анализ функции двух переменных.
    """
    # 1. ОДЗ
    domain = get_domain(f, (x, y))
    display(Math(rf"\text{{Область определения: }} {domain}"))
    
    # Определяем ограничения для графиков
    limits = get_graph_limits([domain], (x, y))
    y_min = limits.get('y_min', y_range[0])
    y_max = limits.get('y_max', y_range[1])
    x_min = limits.get('x_min', x_range[0])
    x_max = limits.get('x_max', x_range[1])
    
    # 2. Частные производные
    fx = diff(f, x)
    fy = diff(f, y)
    
    display(Math(f"f(x,y) = {latex(f)}"))
    display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = {latex(fx)}"))
    display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = {latex(fy)}"))
    
    # 3. Нули производных (раздельно)
    print("\n" + "=" * 60)
    print("НУЛИ ЧАСТНЫХ ПРОИЗВОДНЫХ")
    print("=" * 60)
    
    try:
        zeros_x = solve(fx, (x, y))
    except:
        # Если не удаётся решить систему, решаем по отдельности
        zeros_x = solve(fx, x) or solve(fx, y)
    
    try:
        zeros_y = solve(fy, (x, y))
    except:
        zeros_y = solve(fy, x) or solve(fy, y)
    
    if zeros_x:
        display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = 0 \\rightarrow {latex(zeros_x)}"))
    else:
        display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = 0 \\rightarrow \\text{{не удалось найти аналитически}}"))
    
    if zeros_y:
        display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = 0 \\rightarrow {latex(zeros_y)}"))
    else:
        display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = 0 \\rightarrow \\text{{не удалось найти аналитически}}"))
    
    # 4. 3D графики
    print("\n" + "=" * 60)
    print("3D ГРАФИКИ ЧАСТНЫХ ПРОИЗВОДНЫХ")
    print("=" * 60)
    
    # Определяем диапазон для y (с учётом ОДЗ)
    if y_min > y_range[0] or y_max < y_range[1]:
        y_range_plot = (max(y_min, y_range[0]), min(y_max, y_range[1]))
    else:
        y_range_plot = y_range
    
    print("\n1. ∂f/∂x =", fx)
    plot3d(fx, (x, x_range[0], x_range[1]), (y, y_range_plot[0], y_range_plot[1]),
           title=get_title(fx, 'fx'),
           xlabel='x', ylabel='y', zlabel='∂f/∂x')
    
    print("\n2. ∂f/∂y =", fy)
    plot3d(fy, (x, x_range[0], x_range[1]), (y, y_range_plot[0], y_range_plot[1]),
           title=get_title(fy, 'fy'),
           xlabel='x', ylabel='y', zlabel='∂f/∂y')
    
    # 5. Графики нулей
    print("\n" + "=" * 60)
    print("ГРАФИКИ НУЛЕЙ НА ПЛОСКОСТИ")
    print("=" * 60)
    
    if zeros_x:
        plot_implicit(Eq(fx, 0), (x, x_range[0], x_range[1]), (y, y_range_plot[0], y_range_plot[1]),
                      title=get_title(fx, 'zeros'),
                      xlabel='x', ylabel='y')
    else:
        print("График нулей ∂f/∂x не построен (нет аналитического решения)")
    
    if zeros_y:
        plot_implicit(Eq(fy, 0), (x, x_range[0], x_range[1]), (y, y_range_plot[0], y_range_plot[1]),
                      title=get_title(fy, 'zeros'),
                      xlabel='x', ylabel='y')
    else:
        print("График нулей ∂f/∂y не построен (нет аналитического решения)")
    
    # 6. Контурные графики
    print("\n" + "=" * 60)
    print("КОНТУРНЫЕ ГРАФИКИ")
    print("=" * 60)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    plot_contour_derivative(fx, x, y, 
                           x_range=(x_range[0], x_range[1]),
                           y_range=(y_range_plot[0], y_range_plot[1]),
                           ax=axes[0])
    plot_contour_derivative(fy, x, y,
                           x_range=(x_range[0], x_range[1]),
                           y_range=(y_range_plot[0], y_range_plot[1]),
                           ax=axes[1])
    
    plt.tight_layout()
    plt.show()
    
    # 7. Итоговый вывод
    print("\n" + "=" * 60)
    print("ИТОГОВЫЙ РЕЗУЛЬТАТ")
    print("=" * 60)
    display(Math(f"f(x,y) = {latex(f)}"))
    display(Math(f"\\frac{{\\partial f}}{{\\partial x}} = {latex(fx)}"))
    display(Math(f"\\frac{{\\partial f}}{{\\partial y}} = {latex(fy)}"))
    display(Math(rf"\text{{ОДЗ: }} {domain}"))


# ============================================
# ЗАПУСК
# ============================================
x, y = symbols('x y')
f = x**3 - 2 * x * y**3

analyze_function(f, x, y, x_range=(-5, 5), y_range=(-5, 5))
```

```python
from sympy import *
from IPython.display import display, Math
import numpy as np

init_printing(use_unicode=False, wrap_line=False, no_global=True)

def linear_regression(data, x1_new=None, x2_new=None, verbose=True):
    """
    Универсальная линейная регрессия y = a0 + a1*x1 + a2*x2
    """
    # Символьные переменные
    a0, a1, a2 = symbols('a0 a1 a2', real=True)
    x1, x2 = symbols('x1 x2', real=True)
    
    # 1. Строим MSE (автоматически для любого количества точек)
    n = len(data)
    MSE = sum((a0 + a1*x1i + a2*x2i - yi)**2 for x1i, x2i, yi in data) / n
    
    # 2. Частные производные
    grad = [diff(MSE, var) for var in (a0, a1, a2)]
    
    # 3. Решение системы (универсально!)
    try:
        sol = solve(grad, (a0, a1, a2), dict=True)
        if sol and len(sol) > 0:
            sol_dict = sol[0]
            a0_opt = sol_dict[a0]
            a1_opt = sol_dict[a1]
            a2_opt = sol_dict[a2]
            method = "ТОЧНОЕ РЕШЕНИЕ (sympy)"
        else:
            raise ValueError("Нет решения")
    except:
        # Если не получилось — используем численное решение
        X = np.array([[1, x1i, x2i] for x1i, x2i, _ in data], dtype=float)
        y = np.array([yi for _, _, yi in data], dtype=float)
        coeffs, _, _, _ = np.linalg.lstsq(X, y, rcond=None)
        a0_opt, a1_opt, a2_opt = coeffs
        sol_dict = {a0: a0_opt, a1: a1_opt, a2: a2_opt}
        method = "ЧИСЛЕННОЕ РЕШЕНИЕ (numpy)"
    
    # 4. Модель
    model = a0_opt + a1_opt*x1 + a2_opt*x2
    
    # 5. Вывод
    if verbose:
        print("=" * 60)
        print(f"ЛИНЕЙНАЯ РЕГРЕССИЯ ({method})")
        print("=" * 60)
        print(f"Количество точек: {n}")
        
        # Вывод MSE
        display(Math(f"MSE = {latex(MSE)}"))
        
        # Частные производные
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_0}} = {latex(simplify(grad[0]))}"))
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_1}} = {latex(simplify(grad[1]))}"))
        display(Math(f"\\frac{{\\partial MSE}}{{\\partial a_2}} = {latex(simplify(grad[2]))}"))
        
        # Коэффициенты
        display(Math(f"y = {latex(model.evalf() if method != 'ТОЧНОЕ РЕШЕНИЕ (sympy)' else model)}"))
        display(Math(f"a_0 = {float(a0_opt):.6f}, a_1 = {float(a1_opt):.6f}, a_2 = {float(a2_opt):.6f}"))
        
        # MSE в точке минимума
        y_pred = np.array([float(a0_opt + a1_opt*x1i + a2_opt*x2i) for x1i, x2i, _ in data])
        y_true = np.array([yi for _, _, yi in data])
        mse = np.mean((y_true - y_pred)**2)
        display(Math(f"MSE_{{\\text{min}}} = {mse:.6f}"))
        
        # Проверка
        print("\nПроверка на исходных данных:")
        for x1i, x2i, yi in data:
            y_pred_i = float(a0_opt + a1_opt*x1i + a2_opt*x2i)
            print(f"  x1={x1i:4d}, x2={x2i:3d} → y={y_pred_i:.4f}, факт={yi:.1f}, ошибка={abs(y_pred_i-yi):.4f}")
        
        # Предсказание
        if x1_new is not None and x2_new is not None:
            y_pred_new = float(a0_opt + a1_opt*x1_new + a2_opt*x2_new)
            print(f"\nПредсказание для x1={x1_new}, x2={x2_new}:")
            display(Math(f"\\hat{{y}} = {y_pred_new:.4f}"))
        
        print("=" * 60)
    
    return model, sol_dict
```
