## 1.1 Введение

- Умножение дробей
$$
\frac{a}{b} * \frac{c}{d} = \frac{ac}{bd}
$$

- Деление дробей
$$
\frac{a}{b} : \frac{c}{d} = \frac{ad}{bc}
$$

- Сложение/вычитание дробей
$$
\frac{a}{b} ± \frac{c}{d} = \frac{ad}{bd} ± \frac{cb}{db} = \frac{ad ± cb}{bd}
$$

## 1.2 Основные операции с дробями

```python
from sympy import Rational

# a = Rational(5, 24)
# b = Rational(7, 36)
# c = Rational(7, 36)

# Умножение дроби на число
print('===================Умножение дроби на число=================')
a = Rational(2, 5)
n = 3
print(f'Умножение {a} на {n}: {a * n}')   # 6/5

# Умножение дробей
print('===================Умножение дробей=================')
a = Rational(3, 2)
b = Rational(5, 6)
print(f'Умножение {a} на {b}: {a * b}')    # 5/4

# Деление дробей
print('===================Деление дробей=================')
a = Rational(3, 2)
b = Rational(5, 6)
print(f'Деление {a} на {b}: {a / b}')    # 9/5

# Сложение дробей с одинаковыми знаменателями
print('===================Сложение дробей с одинаковыми знаменателями=================')
a = Rational(2, 3)
b = Rational(4, 3)
print(f'Сложение {a} и {b}: {a + b}')    # 2

# Вычитание дробей с одинаковыми знаменателями
print('===================Вычитание дробей дробей с одинаковыми знаменателями=================')
a = Rational(8, 3)
b = Rational(11, 3)
print(f'Вычитание {a} и {b}: {a - b}')    # -1

# Сложение дробей с разными знаменателями
print('===================Сложение дробей с разными знаменателями=================')
a = Rational(2, 3)
b = Rational(5, 7)
print(f'Сложение {a} и {b}: {a + b}')    # 29/21

a = Rational(1, 2)
b = Rational(3, 4)
print(f'Сложение {a} и {b}: {a + b}')    # 5/4

# Вычитание дробей с разными знаменателями
print('===================Вычитание дробей дробей с разными знаменателями=================')
a = Rational(2, 5)
b = Rational(6, 9)
print(f'Вычитание {a} и {b}: {a - b}')    # -4/15
```

## 1.4 Приведение подобных слагаемых

```python
from sympy import *
from IPython.display import display, Math

x, y, z, m, n = symbols('x y z m n')

print("=== Символьное упрощение выражений ===\n")

# Пример 1: приведение подобных
expr1 = 2*x + 5*x
result1 = simplify(expr1)
print("Пример 1: сложение подобных")
# pprint(f'2*x + 5*x = {result1}')
display(Math(f"2*x + 5*x = {latex(result1)}"))
print()

# Пример 2: с несколькими переменными
expr2 = -6*x*y + 25*x*y
result2 = simplify(expr2)
print("Пример 2: с произведением переменных")
# pprint(f'-6*x*y + 25*x*y = {result2}')
display(Math(f"-6*x*y + 25*x*y = {latex(result2)}"))
print()

# Пример 3: с дробным коэффициентом (используем Rational)
expr3 = 3*x*y*z + Rational(2, 3)*x*y*z
result3 = simplify(expr3)
print("Пример 3: с дробным коэффициентом (2/3)")
# pprint(f'3*x*y*z + 2/3*x*y*z = {result3}')
display(Math(f"3*x*y*z + 2/3*x*y*z = {latex(result3)}"))
print(f"Численно: {result3.evalf()}")
print()

# Пример 4: с общим множителем (дробь)
expr4 = 33*((x+m)/n) - Rational(28, 3)*((x+m)/n)
result4 = simplify(expr4)
print("Пример 4: вынос общего множителя (x+m)/n")
# pprint(f'33*((x+m)/n) - (28/3)*((x+m)/n) = {result4}')
display(Math(f"33*((x+m)/n) - (28/3)*((x+m)/n) = {latex(result4)}"))
print(f"Проверка: 33 - 28/3 = {Rational(33,1) - Rational(28,3)}")
print()

# Дополнительный пример: автоматический общий знаменатель
print("=== Дополнительно: приведение к общему знаменателю ===")
expr5 = 1/x + 1/y
result5 = simplify(expr5)
# pprint(f'1/x + 1/y = {result5}')
display(Math(f"31/x + 1/y = {latex(result5)}"))
```

```python
from sympy import *
from IPython.display import display, Math


# init_printing()

x, y = symbols('x y')

# Исходное выражение
expr = (x + y)**2
print("Исходное выражение:")
# print(expr)
display(Math(latex(expr)))

# РАСКРЫТИЕ СКОБОК (разложение)
print("\n=== Раскрытие скобок (expand) ===")
expanded = expand(expr)
print("Результат:")
# print(expanded)
display(Math(latex(expanded)))

# ОБРАТНОЕ ДЕЙСТВИЕ (разложение на множители)
print("\n=== Разложение на множители (factor) ===")
factored = factor(expanded)
print("Результат:")
# print(factored)
display(Math(latex(factored)))
```

### Задание сложного уровня

Упростите выражение 

$$
\frac{3x^2 + 4x}{x^2 - 2x} - \frac{2x + 7}{x} - \frac{x + 8}{x - 2}
$$

#### Решение от руки

$
\frac{3x^2 + 4x}{x^2 - 2x} - \frac{2x + 7}{x} - \frac{x + 8}{x - 2} = 
\frac{x(3x + 4)}{x(x - 2)} - \frac{2x + 7}{x} - \frac{x + 8}{x - 2} = 
\frac{2x - 4}{(x - 2)} - \frac{2x + 7}{x} = 2 - 2 - \frac{7}{x} = -\frac{7}{x}
$

#### Решение в SymPy

```python
from sympy import *
x = Symbol('x')
expr = (3*x**2 + 4*x)/(x**2 - 2*x) - (2*x + 7)/x - (x+8)/(x-2)
expr
```

```python
expr = expr.simplify()
expr
```

```python
# Улучшенный вариант с явным учётом ОДЗ
from sympy import symbols, simplify, solveset, S, init_printing

# Включаем красивое отображение
init_printing()

x = symbols('x', real=True)
expr = (3*x**2 + 4*x)/(x**2 - 2*x) - (2*x + 7)/x - (x + 8)/(x - 2)

simplified = simplify(expr)

# ОДЗ
denominators = [x**2 - 2*x, x, x - 2]
domain = S.Reals
for den in denominators:
    domain -= solveset(den, x, domain)

# Вывод без print — просто указываем объект
display(simplified)    # или simplified
display(domain)
```

```python
from sympy import symbols, simplify, solveset, S, init_printing, pprint

init_printing(use_unicode=True, pretty_print=True)

x = symbols('x', real=True)
expr = (3*x**2 + 4*x)/(x**2 - 2*x) - (2*x + 7)/x - (x + 8)/(x - 2)

simplified = simplify(expr)

denominators = [x**2 - 2*x, x, x - 2]
domain = S.Reals
for den in denominators:
    domain -= solveset(den, x, domain)

print("Упрощённое выражение:")
pprint(simplified)

print("\nОДЗ:")
pprint(domain)

print(f"\nОтвет: {simplified}, x ∈", end=" ")
pprint(domain)
```

```python
from sympy import symbols, simplify, solveset, S, init_printing, pprint, denom, factor
import matplotlib.pyplot as plt
import numpy as np

init_printing(use_unicode=True, pretty_print=True)

# ============================================
# УЛУЧШЕННАЯ ФУНКЦИЯ ДЛЯ ПОИСКА ЗАПРЕЩЁННЫХ ТОЧЕК
# ============================================

def find_forbidden_points(expr, x):
    """Находит все значения x, при которых хотя бы один знаменатель обращается в ноль"""
    from sympy import fraction, factor, Mul
    
    forbidden = S.EmptySet
    
    # Рекурсивно ищем все знаменатели в выражении
    def get_denominators(expr_):
        dens = []
        if expr_.is_Mul or expr_.is_Add:
            for arg in expr_.args:
                dens.extend(get_denominators(arg))
        elif expr_.is_Pow and expr_.exp < 0:
            dens.append(expr_.base)
        elif expr_.is_Rational and expr_.q != 1:
            # Для чисел-дробей знаменатель — константа, не даёт ограничений
            pass
        else:
            num, den = fraction(expr_)
            if den != 1:
                dens.append(den)
        return dens
    
    denominators = get_denominators(expr)
    
    # Добавляем также знаменатель после приведения к общему виду
    expr_common = expr.cancel()
    num, den = fraction(expr_common)
    if den != 1:
        denominators.append(den)
    
    # Объединяем и решаем
    from sympy import prod
    if denominators:
        # Перемножаем все знаменатели и раскладываем на множители
        total_denom = prod(denominators)
        total_denom = factor(total_denom)
        forbidden = solveset(total_denom, x, domain=S.Reals)
    
    return forbidden


def analyze_rational_expression(expr, x, x_range=(-5, 5), plot=True):
    """Анализ рационального выражения с автоматическим поиском ОДЗ"""
    
    print("=" * 70)
    print("АНАЛИЗ РАЦИОНАЛЬНОГО ВЫРАЖЕНИЯ")
    print("=" * 70)
    
    print("\n1. Исходное выражение:")
    pprint(expr)
    
    # 2. Упрощение
    simplified = simplify(expr)
    print("\n2. Упрощённое выражение:")
    pprint(simplified)
    
    # 3. АВТОМАТИЧЕСКИЙ ПОИСК ЗАПРЕЩЁННЫХ ТОЧЕК
    print("\n3. Область допустимых значений (ОДЗ):")
    
    forbidden_points_set = find_forbidden_points(expr, x)
    domain = S.Reals - forbidden_points_set
    
    # Преобразуем в список для удобства
    forbidden_list = []
    for pt in forbidden_points_set:
        if pt.is_real:
            try:
                forbidden_list.append(float(pt))
            except:
                forbidden_list.append(pt)
    
    print(f"   Запрещённые точки (знаменатель = 0): {forbidden_list}")
    print(f"   ОДЗ: {domain}")
    
    # 4. Разложение знаменателя на множители
    print("\n4. Разложение знаменателя на множители:")
    expr_common = expr.cancel()
    num, den = fraction(expr_common)
    if den != 1:
        from sympy import factor
        factored_den = factor(den)
        print(f"   {den} = {factored_den}")
    
    # 5. Ответ
    print("\n5. ОТВЕТ:")
    print(f"   Упрощённое выражение: {simplified}")
    print(f"   ОДЗ: x ∈ {domain}")
    
    # 6. Визуализация
    if plot:
        visualize_odz(domain, forbidden_list, x_range)
        plot_function(expr, simplified, forbidden_list, x, x_range)
    
    return simplified, domain


def visualize_odz(domain, forbidden_points, x_range=(-5, 5)):
    """Визуализация ОДЗ на числовой прямой"""
    print("\n" + "=" * 70)
    print("ВИЗУАЛИЗАЦИЯ ОДЗ")
    print("=" * 70)
    
    fig, ax = plt.subplots(figsize=(12, 2.5))
    fig.patch.set_facecolor('white')
    
    x_min, x_max = x_range
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(-0.5, 1)
    ax.set_yticks([])
    ax.spines['top'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_position('zero')
    
    # Закрашиваем интервалы ОДЗ
    # Определяем границы интервалов
    points = sorted([p for p in forbidden_points if x_min <= p <= x_max])
    intervals = []
    
    # Добавляем интервалы между точками
    prev = x_min
    for p in points:
        if prev < p:
            intervals.append((prev, p))
        prev = p
    if prev < x_max:
        intervals.append((prev, x_max))
    
    # Закрашиваем интервалы
    for left, right in intervals:
        ax.axvspan(left, right, alpha=0.25, color='green', zorder=1)
    
    # Отмечаем запрещённые точки
    for point in forbidden_points:
        if x_min <= point <= x_max:
            ax.plot(point, 0, 'rx', markersize=12, markeredgewidth=2.5, zorder=5)
            ax.annotate(f'$x = {point}$', xy=(point, 0), xytext=(point, -0.35),
                        ha='center', fontsize=10, color='red', fontweight='bold')
    
    # Легенда
    ax.plot([], [], 'rx', markersize=10, label='Запрещённые точки (разрывы)')
    ax.fill_between([], [], [], color='green', alpha=0.25, label='ОДЗ (разрешённые значения)')
    ax.legend(loc='upper right', fontsize=9)
    
    ax.set_title(f'Область допустимых значений (ОДЗ): {domain}', fontsize=12, pad=20)
    
    plt.tight_layout()
    plt.show()


def plot_function(expr, simplified, forbidden_points, x, x_range=(-5, 5)):
    """Построение графика функции"""
    print("\n" + "=" * 70)
    print("ГРАФИК ФУНКЦИИ")
    print("=" * 70)
    
    from sympy import lambdify
    
    f_numeric = lambdify(x, simplified, 'numpy')
    
    x_min, x_max = x_range
    x_vals_all = np.linspace(x_min, x_max, 1000)
    
    # Маска для исключения запрещённых точек
    mask = np.ones(len(x_vals_all), dtype=bool)
    for point in forbidden_points:
        mask &= np.abs(x_vals_all - point) > 0.01
    
    x_vals = x_vals_all[mask]
    
    try:
        y_vals = f_numeric(x_vals)
    except:
        y_vals = np.zeros_like(x_vals)
    
    plt.figure(figsize=(12, 7))
    plt.plot(x_vals, y_vals, 'b-', linewidth=2.5, label=f'$f(x) = {simplified}$')
    
    # Вертикальные асимптоты
    for i, point in enumerate(forbidden_points):
        if x_min <= point <= x_max:
            plt.axvline(x=point, color='red', linestyle='--', linewidth=1.5,
                       label=f'Асимптота $x = {point}$' if i == 0 else "")
            plt.scatter(point, 0, c='red', s=100, marker='x', zorder=5)
    
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.xlabel('x', fontsize=12)
    plt.ylabel('f(x)', fontsize=12)
    plt.title(f'График функции $f(x) = {simplified}$', fontsize=12)
    plt.legend(loc='best', fontsize=10)
    plt.xlim(x_min, x_max)
    
    valid_y = y_vals[np.isfinite(y_vals)]
    if len(valid_y) > 0:
        y_lim = np.percentile(np.abs(valid_y), 95)
        plt.ylim(-y_lim*1.1, y_lim*1.1)
    else:
        plt.ylim(-10, 10)
    
    plt.tight_layout()
    plt.show()


# ============================================
# ЗАПУСК
# ============================================

x = symbols('x', real=True)

expr = 5/(x+3) + 2/(x-2) - (3*x + 1)/(x**2 + x - 6)
analyze_rational_expression(expr, x, x_range=(-5, 5))
```

\frac{x}{x - y} - \frac{y}{x + y} + \frac{2xy}{x^2 - y^2} = \frac{x}{x - y} - \frac{y}{x + y} + \frac{2xy}{(x-y)(x+y)} = 
$$
$$
\frac{x(x+y) - y(x-y) + 2xy}{(x-y)(x+y)} = \frac{x^2+xy-xy+y^2+2xy}{(x-y)(x+y)} = \frac{x^2+2xy+y^2}{(x-y)(x+y)} =
$$
$$
\frac{(x+y)^2}{(x-y)(x+y)} = \frac{x+y}{(x-y)}
$$
