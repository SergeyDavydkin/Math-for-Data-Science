Реализуйте функцию на языке Python, принимающую на вход n и возвращающую значение следующей суммы:

$$\sum_{i=1}^{n} \ln{i}$$

```python
def ln_sum(n: int):
    """
    Функция  возвращает  кортеж  из значения и 
    выражения суммы логарифмов чисел от 1 до n
    """
    expr = UnevaluatedExpr(log(1))
    for i in range(2, n + 1):
        expr += UnevaluatedExpr(log(i))
    return expr.simplify().evalf()
```

```python
from sympy import log, UnevaluatedExpr, summation, symbols

def ln_sum(n: int):
    """
    Функция возвращает кортеж из:
    - символьного выражения суммы логарифмов от 1 до n
    - численного значения этой суммы
    """
    # Создаём символьную переменную
    i = symbols('i')
    
    # Строим символьное выражение суммы
    expr = summation(log(i), (i, 1, n))
    
    # Вычисляем численное значение
    numeric_value = expr.evalf()
    
    return expr, numeric_value

# Пример использования
expr, value = ln_sum(10)
print(f"Символьное выражение: {expr}")
print(f"Численное значение: {value}")
```

```python
# Улучшенный гибридный вариант
from sympy import log, summation, symbols, N

def ln_sum(n: int, symbolic=True):
    """
    Если symbolic=True — возвращает символьное выражение
    Если False — быстрое численное вычисление
    """
    if symbolic and n <= 1000:  # Для малых n — символьный режим
        i = symbols('i')
        expr = summation(log(i), (i, 1, n))
        return expr, expr.evalf()
    else:  # Для больших n — численный расчёт
        import math
        value = math.fsum(math.log(i) for i in range(1, n + 1))
        return None, value

# Примеры
print(ln_sum(5, symbolic=True))   # (log(120), 4.78749174278205)
print(ln_sum(100000, symbolic=False))  # (None, ~1051299.9)
```

```python
# Исправленный код
from sympy import UnevaluatedExpr, log, simplify, N

def product_ln(n: int):
    """
    Функция возвращает кортеж из:
    - символьного выражения ln(∏ i)
    - численного значения
    """
    # Строим произведение 1·2·...·n
    expr = UnevaluatedExpr(1)
    for i in range(2, n + 1):
        expr *= UnevaluatedExpr(i)
    
    # Берём логарифм
    log_expr = log(expr)
    
    # Упрощаем (должно получиться log(factorial(n)))
    simplified = simplify(log_expr)
    
    # Численное значение
    numeric = simplified.evalf()
    
    return simplified, numeric

# Пример использования
expression, value = product_ln(10)
print(f"Выражение: {expression}")
print(f"Значение: {value}")
```

```python
from sympy import UnevaluatedExpr, log, simplify, N

def product_ln(n: int):
    expr = UnevaluatedExpr(1)
    for i in range(2, n + 1):
        expr *= UnevaluatedExpr(i)
    log_expr = log(expr)
    return simplify(log_expr), N(log_expr)

# Использование
expression, value = product_ln(10)
print(f"ln(10!) = {expression} = {value}")
# ln(10!) = log(3628800) = 15.1044125730755
```

```python
# Улучшенный гибридный вариант
import math
from sympy import UnevaluatedExpr, log, simplify, expand_log

def product_ln(n: int, as_sum: bool = False):
    # Строим произведение
    prod = UnevaluatedExpr(1)
    for i in range(2, n + 1):
        prod *= UnevaluatedExpr(i)
    
    # Логарифм произведения
    log_expr = log(prod)
    
    # Выбираем представление
    if as_sum:
        log_expr = expand_log(log_expr, force=True)
    else:
        log_expr = simplify(log_expr)
    
    # Вычисляем числовое значение через math (без SymPy)
    numeric_value = math.log(math.factorial(n)) if n >= 1 else 0.0
    
    return log_expr, numeric_value

# Примеры
for n in [1, 2, 3, 5, 10]:
    expr_sum, val_sum = product_ln(n, as_sum=True)
    expr_prod, val_prod = product_ln(n, as_sum=False)
    
    print(f"\nn = {n}")
    print(f"  Сумма: {expr_sum} = {val_sum:.6f}")
    print(f"  Произведение: {expr_prod} = {val_prod:.6f}")
```

```python
import math
from sympy import UnevaluatedExpr, log, simplify, expand_log, N, re

def ln_sum(n: int):
    """Сумма логарифмов: Σ ln(i) от i=1 до n"""
    expr = UnevaluatedExpr(0)
    for i in range(1, n + 1):
        expr += UnevaluatedExpr(log(i))
    # Упрощаем и берём действительную часть
    simplified = simplify(expr)
    # Численное значение через evalf()
    numeric = N(simplified)
    return simplified, numeric

def product_ln(n: int, as_sum: bool = False):
    """Логарифм произведения: ln(∏ i) от i=1 до n"""
    prod = UnevaluatedExpr(1)
    for i in range(1, n + 1):
        prod *= UnevaluatedExpr(i)
    log_expr = log(prod)
    if as_sum:
        log_expr = expand_log(log_expr, force=True)
    else:
        log_expr = simplify(log_expr)
    # Численное значение через evalf()
    numeric = N(log_expr)
    return log_expr, numeric

# Сравнение результатов
print("=== Сравнение ln_sum() и product_ln() ===\n")

for n in [1, 2, 3, 5, 10]:
    # Вычисляем обоими способами
    expr_sum, val_sum = ln_sum(n)
    expr_prod, val_prod = product_ln(n)
    
    # Сравниваем символьные выражения
    # Используем simplify с действительной частью
    diff = simplify(expr_sum - expr_prod)
    # Если разность упрощается до 0 или близка к 0
    is_equal = diff == 0 or N(diff).is_zero
    
    print(f"n = {n}")
    print(f"  ln_sum({n})     = {expr_sum}")
    print(f"    Численно: {val_sum:.6f}")
    print(f"  product_ln({n}) = {expr_prod}")
    print(f"    Численно: {val_prod:.6f}")
    print(f"  Эквивалентны: {is_equal}")
    print()
```

Почему это важно в Data Science
|Применение	|Почему используется сумма логарифмов
|-----------|-------------------------------------
|MLE (Maximum Likelihood Estimation)	|Вероятности очень маленькие → произведение → переполнение. Берём логарифм → сумма
|Логистическая регрессия	|Функция потерь — это сумма логарифмов вероятностей
|Энтропия	| $H = −∑piln(pi)$ — сумма логарифмов
|Градиентный спуск	|Производная суммы логарифмов вычисляется проще

$$
\displaystyle\sum_{i=1}^n \frac{1}{i}
$$

```python
def int_sum(n: int):
    result = 0.0
    for i in range(1, n+1):
        result += (1 / i)
    return result
```

#### Задание 6
Реализуйте функцию на языке Python, принимающую на вход n и возвращающую значение следующего произведения:
$$
\prod_{i=1}^n (1+\frac{1}{i})
$$
Проверьте, что для n≥1 результат равен n+1. Объясните, почему это так.

```python
def product_increment(n: int) -> float:
    """
    Возвращает произведение (1 + 1/1) * (1 + 1/2) * ... * (1 + 1/n)
    """
    result = 1.0
    for i in range(1, n + 1):
        result *= 1 + 1 / i
    return result

# Проверка
for n in range(1, 6):
    print(f"n = {n}: {product_increment(n)} = {n + 1}")
```
