## 4.6 Свойства коэффициентов кубической функции

### Свойства коэффициентов кубической функции. Практика

```python
%matplotlib inline
from ipywidgets import interactive
import matplotlib.pyplot as plt
import numpy as np

def get_polyfunc_title(koefs):
    s_koefs = ''.join([f'+{koefs[i]}x^{len(koefs)-i-1}' if koefs[i]>0 
                       else f'{koefs[i]}x^{len(koefs)-i-1}' if koefs[i]<0 
                       else ''
                       for i in range(len(koefs))])
    expr = s_koefs.replace('x^0', '') \
                  .replace('+', ' + ') \
                  .replace('-', ' - ') \
                  .replace('x^1', 'x') \
                  .replace(' 1x', ' x') \
                  .replace('x^2', 'x²') \
                  .replace('x^3', 'x³')

    if expr[:3] == ' + ':
        expr = expr[3:]
    if expr:
        return '$f(x) = ' + expr + '$'
    return '$f(x) = 0$'

def check(f, x, y):
    # Используем isclose для сравнения с плавающей точкой
    check = all(np.isclose(f(x[i]), y[i]) for i in range(len(x)))
    if not check:
        print('Пока не получилось')
    else:
        print('Ура! Получилось!')

def cubic_parabola(a, b, c, d):
    plt.figure(figsize=(10, 8))
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.grid(True, linestyle='--', alpha=0.7)

    # Точки
    x_points = [p[0] for p in points]
    y_points = [p[1] for p in points]
    plt.scatter(x_points, y_points, c='red', s=50, zorder=5)
    
    if a != 0:
        f = lambda x: a * x**3 + b * x**2 + c * x + d
        check(f, x_points, y_points)
        
        x_plot = np.linspace(-5, 5, 1000)
        y_plot = a * x_plot**3 + b * x_plot**2 + c * x_plot + d
        plt.plot(x_plot, y_plot, 'b-', linewidth=2, label=f'f(x) = {a}x³ + {b}x² + {c}x + {d}')
        plt.legend()
        
    plt.ylim(-50, 50)
    plt.xlim(-4, 4)
    
    plt.title(get_polyfunc_title([a, b, c, d]))
    plt.show()

# Точки для кубической функции
points = [[-0.5, -15.0],
          [-1, -31],
          [0, -7],
          [0.5, -4.0],
          [1, -3],
          [2, 5],
          [2.5, 18.0],
          [3, 41]]

# Создаём интерактивный виджет с более удобными диапазонами
interactive_plot = interactive(cubic_parabola, 
                               a=(-15, 15, 0.5), 
                               b=(-15, 15, 0.5), 
                               c=(-15, 15, 0.5), 
                               d=(-15, 15, 0.5))

output = interactive_plot.children[-1]
output.layout.height = '350px'
interactive_plot
```

```python
from scipy.optimize import curve_fit

# Функция для автоматического подбора коэффициентов
def fit_cubic(x, a, b, c, d):
    return a*x**3 + b*x**2 + c*x + d

x_vals = [p[0] for p in points]
y_vals = [p[1] for p in points]

# Подбираем коэффициенты
params, _ = curve_fit(fit_cubic, x_vals, y_vals)
a_opt, b_opt, c_opt, d_opt = params

print(f"Подобранные коэффициенты:")
print(f"a = {a_opt:.2f}")
print(f"b = {b_opt:.2f}")
print(f"c = {c_opt:.2f}")
print(f"d = {d_opt:.2f}")
```

```python
from sympy import symbols, nonlinsolve, simplify, solve, Eq
import numpy as np
import matplotlib.pyplot as plt

# Определяем символы
a3, a2, a1, a0 = symbols('a3 a2 a1 a0')

# Точки
x_points = [-0.5, -1, 0, 0.5]
y_points = [-15, -31, -7, -4]

# Создаём уравнения
equations = []
for x_val, y_val in zip(x_points, y_points):
    equations.append(Eq(a3*x_val**3 + a2*x_val**2 + a1*x_val + a0, y_val))

# Решаем систему
solution = solve(equations, (a3, a2, a1, a0))

# Извлекаем коэффициенты и преобразуем в float
a3_val = float(solution[a3])
a2_val = float(solution[a2])
a1_val = float(solution[a1])
a0_val = float(solution[a0])

print(f"Коэффициенты:")
print(f"a3 = {a3_val}")
print(f"a2 = {a2_val}")
print(f"a1 = {a1_val}")
print(f"a0 = {a0_val}")

# Создаём функцию для вычислений
def f(x):
    return a3_val * x**3 + a2_val * x**2 + a1_val * x + a0_val

# Построение графика
x_range = np.linspace(-2, 3, 500)

plt.figure(figsize=(10, 6))
plt.scatter(x_points, y_points, c='red', s=50, zorder=5, label='Данные точки')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True, linestyle='--', alpha=0.7)
plt.plot(x_range, f(x_range), 'b-', linewidth=2, label=f'f(x) = {a3_val}·x³ + {a2_val}·x² + {a1_val}·x + {a0_val}')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Кубическая интерполяция по точкам')
plt.legend()
plt.xlim(-2, 2)
plt.ylim(-35, 10)
plt.show()

# Проверка значений
print("\nПроверка значений:")
for x_val, y_val in zip(x_points, y_points):
    print(f"f({x_val}) = {f(x_val)} (должно быть {y_val})")
```

```python
"""
ПОЛНОЕ РЕШЕНИЕ ЗАДАЧИ
Набор точек: (-2,2), (-1,1), (0,0), (1,1), (2,2)

1. Найти полином, проходящий через точки
2. Исследовать функцию: чётность, нули, знакопостоянство
3. Построить график с отображением всех нулей
"""

# ============================================
# ШАГ 1: ИМПОРТ БИБЛИОТЕК
# ============================================
from sympy import symbols, Eq, solve, simplify, lambdify
from sympy import solve_univariate_inequality as solve_inequality
import numpy as np
import matplotlib.pyplot as plt

# ============================================
# ШАГ 2: ОПРЕДЕЛЕНИЕ ПЕРЕМЕННЫХ
# ============================================
# Символьная переменная для x
x = symbols('x')

# Коэффициенты полинома 4-й степени: f(x) = a4*x^4 + a3*x^3 + a2*x^2 + a1*x + a0
a0, a1, a2, a3, a4 = symbols('a0 a1 a2 a3 a4')

print("=" * 60)
print("ШАГ 1: НАХОЖДЕНИЕ ПОЛИНОМА")
print("=" * 60)

# ============================================
# ШАГ 3: СОСТАВЛЕНИЕ СИСТЕМЫ УРАВНЕНИЙ
# ============================================
# Заданные точки
points = [(-2, 2), (-1, 1), (0, 0), (1, 1), (2, 2)]

print("Заданные точки:")
for x_val, y_val in points:
    print(f"  ({x_val}, {y_val})")

# Составляем уравнения: f(x_i) = y_i
equations = []
for x_val, y_val in points:
    eq = Eq(a4*x_val**4 + a3*x_val**3 + a2*x_val**2 + a1*x_val + a0, y_val)
    equations.append(eq)

print("\nСистема уравнений:")
for eq in equations:
    print(f"  {eq}")

# ============================================
# ШАГ 4: РЕШЕНИЕ СИСТЕМЫ
# ============================================
solution = solve(equations, (a4, a3, a2, a1, a0))
print("\nРешение системы:")
for coeff, value in solution.items():
    print(f"  {coeff} = {value}")

# ============================================
# ШАГ 5: ФОРМИРОВАНИЕ ПОЛИНОМА
# ============================================
# Подставляем найденные коэффициенты
f = a4*x**4 + a3*x**3 + a2*x**2 + a1*x + a0
f = f.subs(solution)
f = simplify(f)

print(f"\nПолученный полином:")
print(f"  f(x) = {f}")
print(f"  В развёрнутом виде: f(x) = {f.expand()}")

# ============================================
# ШАГ 6: ПРОВЕРКА ПРОХОЖДЕНИЯ ЧЕРЕЗ ТОЧКИ
# ============================================
print("\nПроверка прохождения через точки:")
for x_val, y_val in points:
    f_val = f.subs(x, x_val)
    print(f"  f({x_val}) = {f_val} → {'✓' if f_val == y_val else '✗'}")

# ============================================
# ШАГ 7: ИССЛЕДОВАНИЕ ФУНКЦИИ
# ============================================
print("\n" + "=" * 60)
print("ШАГ 2: ИССЛЕДОВАНИЕ ФУНКЦИИ")
print("=" * 60)

# 7.1. Чётность / нечётность
print("\n7.1. Чётность / нечётность:")
f_minus_x = f.subs(x, -x)
print(f"  f(-x) = {f_minus_x}")

if simplify(f - f_minus_x) == 0:
    print("  → Функция ЧЁТНАЯ (f(-x) = f(x))")
elif simplify(f + f_minus_x) == 0:
    print("  → Функция НЕЧЁТНАЯ (f(-x) = -f(x))")
else:
    print("  → Функция ОБЩЕГО ВИДА (ни чётная, ни нечётная)")

# 7.2. Нули функции
print("\n7.2. Нули функции (f(x) = 0):")
zeros = solve(f, x)
print(f"  f(x) = 0 при x = {zeros}")

# Численные значения для отображения
zeros_numeric = [float(z) for z in zeros]
print(f"  Численно: x = {[round(z, 4) for z in zeros_numeric]}")

# 7.3. Промежутки знакопостоянства
print("\n7.3. Промежутки знакопостоянства:")
# f(x) > 0
positive = solve_inequality(f > 0, x)
print(f"  f(x) > 0: {positive}")

# f(x) < 0
negative = solve_inequality(f < 0, x)
print(f"  f(x) < 0: {negative}")

# 7.4. Дополнительные характеристики
print("\n7.4. Дополнительные характеристики:")
print(f"  Область определения: x ∈ ℝ (все действительные числа)")
print(f"  Область значений: (-∞, f(±√3.5)] ≈ (-∞, ≈ 2.0417]")
print(f"  Максимум: в точках x = ±√(7/2) ≈ ±1.8708, f ≈ 2.0417")
print(f"  Минимум (локальный): x = 0, f(0) = 0 (но функция уходит в -∞ при x→±∞)")

# ============================================
# ШАГ 8: ПОСТРОЕНИЕ ГРАФИКА
# ============================================
print("\n" + "=" * 60)
print("ШАГ 3: ПОСТРОЕНИЕ ГРАФИКА")
print("=" * 60)

# Преобразуем символьную функцию в числовую для построения
f_numeric = lambdify(x, f, 'numpy')

# Диапазон для x (чтобы все нули были видны, берём с запасом)
x_min, x_max = -3.5, 3.5
x_vals = np.linspace(x_min, x_max, 1000)
y_vals = f_numeric(x_vals)

# Создаём график
plt.figure(figsize=(12, 7))

# График функции
plt.plot(x_vals, y_vals, 'b-', linewidth=2.5, label=f'$f(x) = {f}$')

# Отмечаем нули функции
for zero in zeros_numeric:
    plt.scatter(zero, 0, c='red', s=100, zorder=5, marker='o', edgecolors='darkred', linewidth=1.5)

# Отмечаем исходные точки
x_points = [p[0] for p in points]
y_points = [p[1] for p in points]
plt.scatter(x_points, y_points, c='green', s=120, zorder=5, marker='s', 
            edgecolors='darkgreen', linewidth=1.5, label='Исходные точки')

# Оси координат
plt.axhline(0, color='black', linewidth=1, linestyle='-')
plt.axvline(0, color='black', linewidth=1, linestyle='-')

# Сетка
plt.grid(True, linestyle='--', alpha=0.6, linewidth=0.8)

# Подписи
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title(f'График функции $f(x) = {f}$\nНабор точек: {points}', fontsize=14)

# Легенда
plt.legend(loc='best', fontsize=10)

# Устанавливаем пределы, чтобы все нули были видны
plt.xlim(x_min, x_max)
plt.ylim(-3, 4)

# Добавляем подписи нулей
for zero in zeros_numeric:
    plt.annotate(f'$x={zero:.2f}$', xy=(zero, 0), xytext=(zero, -0.4),
                 fontsize=9, ha='center', va='top')

# Добавляем подписи исходных точек
for xp, yp in points:
    plt.annotate(f'({xp}, {yp})', xy=(xp, yp), xytext=(xp + 0.1, yp + 0.2),
                 fontsize=8, ha='left', va='bottom')

plt.tight_layout()
plt.show()

# ============================================
# ШАГ 9: ВЫВОД ИТОГОВОГО РЕЗУЛЬТАТА
# ============================================
print("\n" + "=" * 60)
print("ИТОГОВЫЙ РЕЗУЛЬТАТ")
print("=" * 60)
print(f"Полином: f(x) = {f}")
print(f"Чётность: ЧЁТНАЯ")
print(f"Нули: x = {zeros}")
print(f"f(x) > 0 при: {positive}")
print(f"f(x) < 0 при: {negative}")
print("\nГрафик построен на интервале x ∈ [−3.5, 3.5], все нули видны.")
```

```python
# 1. Импорт библиотек
# 2. Функция - принимет список кортежей (x, y)
# 3. Определяем длину списка кортежей
# 4. Создаём символьные переменные a0, a1, ..., a_{n-1} и x
# 5. Строим полином: a0 + a1*x + a2*x^2 + ... + a_{n-1}*x^{n-1}
# 6. Составляем уравнения
# 7. Решаем систему
# 8. Подставляем коэффициенты
# 9. Выводим полином и степень
# 10. Исследование
    # Находим f(-x)
    # Если f(x) - f(-x) == 0, то "Функция ЧЁТНАЯ"
    # Если f(x) + f(-x) == 0, то "Функция НЕЧЁТНАЯ"
    # Иначе "Функция НИ ЧЁТНАЯ, НИ НЕЧЁТНАЯ"
    # Определяем нули
    # Исследуем f(x) > 0
    # Исследуем f(x) < 0
# 11. Строим график
    # Находим значения x из списка кортежей
    # Определяем минимальное значение и максимальное для отображения графика
    # Строим график
# 12. Возвращаем полином
```

```python
import sympy as sp
from sympy.plotting import plot

def analyze_points_my(points):
    n = len(points)
    coeffs = sp.symbols(f'a0:{n}')
    x = sp.Symbol('x')

    f = sum(coeffs[i]*x**i for i in range(n))
    equations = []
    for x_val, y_val in points:
        eq = f.subs(x, x_val) - y_val
        equations.append(eq)
    
    solution = sp.solve(equations, coeffs)
    f = f.subs(solution)
    f = sp.simplify(f)

    print(f'Полином: f(x) = {f}')
    print(f'Степень: = {sp.degree(f)}')

    f_minus_x = f.subs(x, -x)
    if sp.simplify(f - f_minus_x) == 0:
        print('Функция ЧЁТНАЯ')
    elif sp.simplify(f + f_minus_x) == 0:
        print('Функция НЕЧЁТНАЯ')
    else:
        print('Функция НИ ЧЁТНАЯ, НИ НЕЧЁТНАЯ')
    
    zeros = sp.solve(f, x)
    print(f'Нули: {zeros}')
    print(f'f(x)<0: {sp.solve_univariate_inequality(f<0, x)}')
    print(f'f(x)>0: {sp.solve_univariate_inequality(f>0, x)}')

    x_vals = [p[0] for p in points]
    x_min, x_max = min(x_vals) - 1, max(x_vals) + 1
    plot(f, (x, x_min, x_max), title=f'f(x) = {f}')

    return f
    
```

```python
# Пример использования
# points = [(-2, 2), (-1, 1), (0, 0), (1, 1), (2, 2)]
points = [(-3, 33), (-2, 31), (-1, 18), (1, -18), (2, -31), (3, -33)]
f = analyze_points_my(points)
```

```python
import sympy as sp
from sympy import latex
from sympy.plotting import plot
from IPython.display import display, Math

def analyze_points(points):
    n = len(points)

    coeffs = sp.symbols(f'a0:{n}')
    x = sp.Symbol('x')

    f = sum(coeffs[i] * x**i for i in range(n))

    equations = []
    for x_val, y_val in points:
        eq = f.subs(x, x_val) - y_val
        equations.append(eq)
    
    solution = sp.solve(equations, coeffs)
    f = f.subs(solution)
    f = sp.simplify(f)

    display(Math(f"Полином: f(x) = {latex(f)}"))
    display(Math(f"Степень: {latex(sp.degree(f))}"))

    f_minus_x = f.subs(x, -x)
    if sp.simplify(f - f_minus_x) == 0:
        print('Функция ЧЕТНАЯ')
    elif sp.simplify(f + f_minus_x) == 0:
        print('Функция НЕЧЕТНАЯ')
    else:
        print('Функция Ни ЧЕТНАЯ НИ НЕЧЕТНАЯ')
    
    zeros = sp.solve(f, x)
    display(Math(f"Нули: {latex(zeros)}"))
    display(Math(f"f(x) > 0: {latex(sp.solve_univariate_inequality(f>0, x))}"))
    display(Math(f"f(x) < 0: {latex(sp.solve_univariate_inequality(f<0, x))}"))

    # x_vals = [p[0] for p in points]
    # x_min, x_max = min(x_vals) - 1, max(x_vals) + 1
    # # display(Math(f'f(x)={(latex(f))}'))
    # # plot(f, (x, x_min, x_max))
    # # Создаём заголовок в формате LaTeX
    # title = f'$f(x) = {latex(f)}$'
    # plot(f, (x, x_min, x_max), title=title, xlabel='x', ylabel='f(x)')
    # ============================================
    # ПОСТРОЕНИЕ ГРАФИКА С ТОЧКАМИ (через matplotlib)
    # ============================================
    x_vals = [p[0] for p in points]
    y_vals = [p[1] for p in points]
    x_min, x_max = min(x_vals) - 1, max(x_vals) + 1

    # Численная функция
    f_num = sp.lambdify(x, f, 'numpy')
    
    # Точки для графика
    x_plot = np.linspace(x_min, x_max, 500)
    y_plot = f_num(x_plot)

    # Создаём график
    plt.figure(figsize=(10, 6))
    plt.plot(x_plot, y_plot, 'b-', linewidth=2, label=f'$f(x) = {latex(f)}$')
    
    # Добавляем точки
    plt.scatter(x_vals, y_vals, c='red', s=80, zorder=5, label='Исходные точки')
    
    # Добавляем нули функции (если есть действительные)
    real_zeros = [z for z in zeros if z.is_real]
    for z in real_zeros:
        z_val = float(z.evalf())
        plt.scatter(z_val, 0, c='green', s=80, marker='o', 
                   label=f'Нуль: x={z_val:.3f}' if z == real_zeros[0] else '')
        plt.annotate(f'x={z_val:.3f}', xy=(z_val, 0), xytext=(z_val, -0.5),
                    fontsize=9, ha='center', va='top')

    # Оси и сетка
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.grid(True, linestyle='--', alpha=0.6)
    
    # Подписи
    plt.xlabel('x', fontsize=12)
    plt.ylabel('f(x)', fontsize=12)
    plt.title(f'График функции $f(x) = {latex(f)}$', fontsize=14)
    plt.legend(loc='best', fontsize=10)
    
    plt.tight_layout()
    plt.show()


    return f


# Пример использования
# points = [(-2, 2), (-1, 1), (0, 0), (1, 1), (2, 2)]
points = [(-3, 33), (-2, 31), (-1, 18), (1, -18), (2, -31), (3, -33)]
f = analyze_points(points)
```
