1. 
$$
f(x) = x^2 + 3x -4
$$

$$
f'(x) = 2x + 3
$$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

from sympy import latex
from IPython.display import display, Math

# Определяем переменную
x = sp.Symbol('x')

# Функция и её производная
f = x**2 + 3*x - 4
f_prime = 2*x + 3

print("=" * 50)
print("ИССЛЕДОВАНИЕ ФУНКЦИИ С ПОМОЩЬЮ ПРОИЗВОДНОЙ")
print("=" * 50)
# print(f"f(x) = {f}")
display(Math(f"f(x) = {latex(f)}"))
# print(f"f'(x) = {f_prime}")
display(Math(f"f'(x) = {latex(f_prime)}"))

# 1. Находим критическую точку (f'(x) = 0)
critical_point = sp.solve(f_prime, x)[0]
# print(f"\n1. Критическая точка: x = {critical_point}")
print('Критическая точка: ')
display(Math(f"x = {latex(critical_point)}"))

# 2. Определяем знаки производной
print("\n2. Знаки производной:")
for test_x in [critical_point - 1, critical_point + 1]:
    sign = sp.sign(f_prime.subs(x, test_x))
    direction = "возрастает" if sign > 0 else "убывает"
    # print(f"   x = {test_x}: f'({test_x}) = {f_prime.subs(x, test_x)} → {direction}")
    display(Math(f"   x = {latex(test_x)}: f'({latex(test_x)}) = {latex(f_prime.subs(x, test_x))} → {direction}"))

# 3. Определяем экстремум
f_double_prime = sp.diff(f_prime, x)
print(f"\n3. Вторая производная: f''(x) = {f_double_prime}")

if f_double_prime.subs(x, critical_point) > 0:
    extremum_type = "минимум"
elif f_double_prime.subs(x, critical_point) < 0:
    extremum_type = "максимум"
else:
    extremum_type = "точка перегиба"

f_crit = f.subs(x, critical_point)
print(f"   В точке x = {critical_point}: {extremum_type}, f({critical_point}) = {f_crit}")

# 4. Вывод поведения функции
print("\n4. Поведение функции:")
print(f"   Убывает на интервале: (-∞, {critical_point})")
print(f"   Возрастает на интервале: ({critical_point}, +∞)")

# 5. График
print("\n5. Построение графика...")

# Численные функции
f_num = sp.lambdify(x, f, 'numpy')
x_vals = np.linspace(-5, 3, 500)
y_vals = f_num(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, 'b-', lw=2, label=f'$f(x) = {f}$')
plt.scatter(critical_point, f_crit, c='red', s=100, zorder=5, 
            label=f'{extremum_type.capitalize()} ({critical_point}, {f_crit})')
plt.axhline(0, color='k', lw=0.8)
plt.axvline(0, color='k', lw=0.8)
plt.axvline(critical_point, color='green', linestyle='--', alpha=0.5, 
            label=f'x = {critical_point} (критическая точка)')
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
title = f'$f(x) = {latex(f)}$'
plt.title(f'График функции f(x) = {title}')
plt.legend()
plt.show()
```

2. 
$$
f(x) = x^3 - 2x^2 + x - 6
$$

$$
f'(x) = 3x^2 - 4x + 1
$$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Определяем переменную
x = sp.Symbol('x')

# Функция и её производная
f = x**3 - 2*x**2 + x - 6
f_prime = 3*x**2 - 4*x + 1

print("=" * 60)
print("ИССЛЕДОВАНИЕ ФУНКЦИИ С ПОМОЩЬЮ ПРОИЗВОДНОЙ")
print("=" * 60)
print(f"f(x) = {f}")
print(f"f'(x) = {f_prime}")

# 1. Находим критические точки (f'(x) = 0)
critical_points = sp.solve(f_prime, x)
print(f"\n1. Критические точки: {critical_points}")

# 2. Преобразуем в числа для удобства
critical_points_float = [float(p.evalf()) for p in critical_points]
critical_points_float.sort()
print(f"   Численно: {critical_points_float}")

# 3. Определяем знаки производной на интервалах
print("\n2. Знаки производной:")

# Интервалы: (-∞, x1), (x1, x2), (x2, +∞)
intervals = [(-float('inf'), critical_points_float[0]),
             (critical_points_float[0], critical_points_float[1]),
             (critical_points_float[1], float('inf'))]

test_points = [critical_points_float[0] - 1,
               (critical_points_float[0] + critical_points_float[1]) / 2,
               critical_points_float[1] + 1]

for i, (interval, test_x) in enumerate(zip(intervals, test_points)):
    sign = sp.sign(f_prime.subs(x, test_x))
    direction = "возрастает" if sign > 0 else "убывает"
    print(f"   На интервале {interval}: f'(x) {'>' if sign > 0 else '<'} 0 → {direction}")

# 4. Определяем экстремумы (по второй производной или по смене знака)
f_double_prime = sp.diff(f_prime, x)
print(f"\n3. Вторая производная: f''(x) = {f_double_prime}")

for x0 in critical_points:
    x0_val = float(x0.evalf())
    f_prime_val = f_prime.subs(x, x0)
    f_double_prime_val = f_double_prime.subs(x, x0)
    f_val = f.subs(x, x0)
    
    if f_double_prime_val > 0:
        extremum_type = "минимум"
    elif f_double_prime_val < 0:
        extremum_type = "максимум"
    else:
        extremum_type = "требуется дополнительный анализ"
    
    print(f"   x = {x0_val}: {extremum_type}, f({x0_val}) = {float(f_val.evalf()):.4f}")

# 5. Вывод поведения функции
print("\n4. Поведение функции:")
print(f"   Возрастает на интервалах: (-∞, {critical_points_float[0]:.4f}) и ({critical_points_float[1]:.4f}, +∞)")
print(f"   Убывает на интервале: ({critical_points_float[0]:.4f}, {critical_points_float[1]:.4f})")

# 6. График
print("\n5. Построение графика...")

f_num = sp.lambdify(x, f, 'numpy')
x_vals = np.linspace(-2, 3, 500)
y_vals = f_num(x_vals)

plt.figure(figsize=(12, 6))

# Основной график
plt.plot(x_vals, y_vals, 'b-', lw=2, label=f'$f(x) = {f}$')

# Отмечаем критические точки
colors = ['red', 'green']
for i, x0 in enumerate(critical_points_float):
    y0 = float(f.subs(x, x0).evalf())
    if i == 0:
        label = f'Максимум ({x0:.3f}, {y0:.3f})'
    else:
        label = f'Минимум ({x0:.3f}, {y0:.3f})'
    plt.scatter(x0, y0, c=colors[i], s=100, zorder=5, label=label)

# Отмечаем точку пересечения с осью y
y_intercept = f.subs(x, 0)
plt.scatter(0, y_intercept, c='purple', s=80, zorder=5, label=f'Пересечение с OY (0, {y_intercept})')

# Оси
plt.axhline(0, color='k', lw=0.8)
plt.axvline(0, color='k', lw=0.8)

# Вертикальные линии в критических точках
for x0 in critical_points_float:
    plt.axvline(x0, color='gray', linestyle='--', alpha=0.5)

plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(f'График функции $f(x) = {f}$')
plt.legend()
plt.show()
```

3. 
$$
f(x) = e^{2x-x^2}
$$

$$
f'(x) = (2 - 2x)e^{2x-x^2}
$$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Определяем переменную
x = sp.Symbol('x')

# Функция и её производная
f = sp.exp(2*x - x**2)
f_prime = (2 - 2*x) * sp.exp(2*x - x**2)

print("=" * 60)
print("ИССЛЕДОВАНИЕ ФУНКЦИИ С ПОМОЩЬЮ ПРОИЗВОДНОЙ")
print("=" * 60)
print(f"f(x) = {f}")
print(f"f'(x) = {f_prime}")

# 1. Находим критические точки (f'(x) = 0)
# Экспонента всегда положительна, поэтому f'(x)=0 при (2-2x)=0
critical_points = sp.solve(2 - 2*x, x)
print(f"\n1. Критические точки: {critical_points}")

critical_points_float = [float(p.evalf()) for p in critical_points]
print(f"   Численно: {critical_points_float}")

# 2. Определяем знаки производной на интервалах
print("\n2. Знаки производной:")

# Интервалы: (-∞, x1), (x1, +∞)
test_points = [critical_points_float[0] - 1, critical_points_float[0] + 1]

for test_x in test_points:
    f_prime_val = f_prime.subs(x, test_x)
    sign = sp.sign(f_prime_val)
    direction = "возрастает" if sign > 0 else "убывает"
    print(f"   x = {test_x}: f'({test_x}) = {f_prime_val.evalf():.4f} → {direction}")

# 3. Определяем экстремум
f_double_prime = sp.diff(f_prime, x)
print(f"\n3. Вторая производная: f''(x) = {sp.simplify(f_double_prime)}")

x0 = critical_points_float[0]
f_double_prime_val = f_double_prime.subs(x, x0)
f_val = f.subs(x, x0)

if f_double_prime_val > 0:
    extremum_type = "минимум"
elif f_double_prime_val < 0:
    extremum_type = "максимум"
else:
    extremum_type = "требуется дополнительный анализ"

print(f"   В точке x = {x0}: {extremum_type}")
print(f"   f({x0}) = {f_val.evalf():.6f}")

# 4. Вывод поведения функции
print("\n4. Поведение функции:")
if extremum_type == "максимум":
    print(f"   Возрастает на интервале: (-∞, {x0})")
    print(f"   Убывает на интервале: ({x0}, +∞)")
else:
    print(f"   Убывает на интервале: (-∞, {x0})")
    print(f"   Возрастает на интервале: ({x0}, +∞)")

# 5. График
print("\n5. Построение графика...")

f_num = sp.lambdify(x, f, 'numpy')
x_vals = np.linspace(-1, 3, 500)
y_vals = f_num(x_vals)

plt.figure(figsize=(12, 6))

# Основной график
plt.plot(x_vals, y_vals, 'b-', lw=2, label=f'$f(x) = e^{{2x-x^2}}$')

# Отмечаем критическую точку
y0 = float(f_val.evalf())
plt.scatter(x0, y0, c='red', s=100, zorder=5, 
           label=f'{extremum_type.capitalize()} ({x0:.3f}, {y0:.3f})')

# Оси
plt.axhline(0, color='k', lw=0.8)
plt.axvline(0, color='k', lw=0.8)

# Вертикальная линия в критической точке
plt.axvline(x0, color='green', linestyle='--', alpha=0.5, label=f'x = {x0}')

plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(f'График функции $f(x) = e^{{2x-x^2}}$')
plt.legend()
plt.show()
```

```python
from sympy import *
from sympy.plotting import plot
from IPython.display import display, Math

def function_analysis(f, diff_f, x_range=None):
    """
    Анализ функции с помощью производной.
    
    Параметры:
    f - исходная функция
    diff_f - производная
    x_range - кортеж (x_min, x_max) для графика, например (-5, 2)
    """
    print("=" * 60)
    print('Исходная функция и её производная:')
    print("=" * 60)
    display(Math(f"f(x) = {latex(f)}"))
    display(Math(f"f'(x) = {latex(diff_f)}"))
    print()

    print("=" * 60)
    print('Находим промежуток возрастания:')
    print("=" * 60)
    display(Math(f"f'(x) > 0 = {latex(solve_univariate_inequality(diff_f > 0, x))}"))
    
    print("=" * 60)
    print('Находим промежуток убывания:')
    print("=" * 60)
    display(Math(f"f'(x) < 0 = {latex(solve_univariate_inequality(diff_f < 0, x))}"))

    print("=" * 60)
    print('Находим нули:')
    print("=" * 60)
    display(Math(f"f'(x) = 0 = {latex(solve(diff_f))}"))

    print("=" * 60)
    print('Строим график:')
    print("=" * 60)
    display(Math(f"f(x) = {latex(f)}"))
    
    # Построение графика с указанным диапазоном
    if x_range is None:
        # Автоматический диапазон на основе нулей функции
        zeros = solve(f, x)
        if zeros:
            x_min = min([float(z.evalf()) for z in zeros if z.is_real]) - 2
            x_max = max([float(z.evalf()) for z in zeros if z.is_real]) + 2
            plot(f, (x, x_min, x_max))
        else:
            plot(f, (x, -10, 10))
    else:
        # Использовать заданный диапазон
        plot(f, (x, x_range[0], x_range[1]))
```

```python
# Пример использования 1
x = Symbol('x')

# Для функции f(x) = x² + 3x - 4
f = x**2 + 3*x - 4
diff_f = 2*x + 3

# Вызов с указанием диапазона
function_analysis(f, diff_f, x_range=(-5, 2))
```

## 6.8 Сигмоида и еще несколько правил вычисления производных

### 1. Сигмоида (логистическая функция)

#### Определение

$$
\sigma(x) = \frac{1}{1 + e^{-x}} = \frac{e^x}{1 + e^{x}}
$$

#### Производная сигмоиды

$$
\sigma'(x) = \sigma(x) \sdot (1 - \sigma(x)) = \frac{e^x}{(1 + e^{x})^2}
$$

#### Вывод производной

$$
\sigma'(x) = \left(\frac{1}{1 + e^{x}}\right)' = \frac{0-1 \sdot (-e^{-x})}{(1+e^{-x})^2} = \frac{e^{-x}}{(1 + e^{-x})^2} = \sigma(x) \sdot (1 - \sigma(x))
$$

|Правило|Формула|Пример
|-------|-------|-------
|Костанта | $(C)' = 0$ | (5)'=0
|Степень | $(x^n)' = nx^{n-1}$ | $(x^5)' = 5x^4$
|Константа * функция | $(C \sdot f)' = C \sdot f'$ | $(3x^2)' = 6x $
|Сумма| $(f + g)' = f' + g'$ | $ (x^2 + x^3)' = 2x + 3x^2 $
|Разность| $(f - g)' = f' - g'$| $(x^3 - x)' = 3x^2 - 1$
|Произведение| $(f \sdot g)' = f'g + fg'$| $(x^2 \sdot e^x)' = 2xe^x + x^2e^x$
|Частное| $\left(\frac{f}{g}\right)' = \frac{f'g - fg'}{g^2}$| $\left(\frac{x}{x+1}\right)' = \left(\frac{1}{(x+1)^2}\right)$
|Сложная функция| $(f(g(x)))' = f'(g(x)) \sdot g'(x) $ | $((x^2 + 1)^3)' = 3(x^2 + 1)^2 \sdot 2x $
|дифференцирования показательной функции| $(a^x)' = a^x\ln(a)$| $(6^x)' = 6^x \ln 6$

```python
import sympy as sp
from sympy.plotting import plot
from IPython.display import display, Math

x = Symbol('x')

print("=" * 60)
print("ВЫЧИСЛЕНИЕ ПРОИЗВОДНЫХ В SymPy")
print("=" * 60)

# 1. Сигмоида
sigma = 1 / (1 + exp(-x))
sigma_prime = diff(sigma, x)
print(f"\n1. Сигмоида:")
display(Math(f"σ(x) = {latex(sigma)}"))
display(Math(f"σ'(x) = {latex(simplify(sigma_prime))}"))

# 2. Степень
f1 = x**5
f1_prime = diff(f1, x)
display(Math(f"\n2. Степень: ({latex(f1)})' = {latex(f1_prime)}"))

# 3. Экспонента
f2 = exp(2*x)
f2_prime = diff(f2, x)
display(Math(f"\n3. Экспонента: ({latex(f2)})' = {latex(f2_prime)}"))

# 4. Логарифм
f3 = ln(x)
f3_prime = diff(f3, x)
display(Math(f"\n4. Логарифм: ({latex(f3)})' = {latex(f3_prime)}"))
# print(f"\n4. Логарифм: (ln x)' = {f3_prime}")

# 5. Тригонометрическая
f4 = sin(x**2)
f4_prime = diff(f4, x)
display(Math(f"\n5. Сложная функция: ({latex(f4)})' = {latex(f4_prime)}"))

# 6. Произведение
f5 = x**2 * exp(x)
f5_prime = diff(f5, x)
display(Math(f"\n6. Произведение: ({latex(f5)})' = {latex(f5_prime)}"))

# 7. Частное
f6 = x / (x + 1)
f6_prime = simplify(diff(f6, x))
display(Math(f"\n7. Частное: ({latex(f6)})' = {latex(f6_prime)}"))
```

## Самостоятельный план освоения производных

### Этап 1. Базовые понятия

#### 1.1. Что такое производная?

* Геометрический смысл: угловой коэффициент касательной

* Физический смысл: мгновенная скорость изменения функции

* Определение через предел:
$$
f'(x) = \lim_{x \to 0} \frac{f(x + h) - f(x)}{h}
$$

#### 1.2. Основные правила дифференцирования

|Правило	|Формула
|-----------|----------
|Константа  |$(C)' = 0$
|Степень    |$(x^n)' = nx^{n-1}$
|Умножение на константу|$(C \sdot f)' = C \sdot f'$
|Сумма/разность|$(f \pm g)' = f' \pm g'$
